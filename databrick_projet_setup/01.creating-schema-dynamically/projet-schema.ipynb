```python
bronze_path = spark.sql('DESCRIBE EXTERNAL LOCATION `bronze-gcp_bucket`').select("url").collect()[0][0]
silver_path = spark.sql('DESCRIBE EXTERNAL LOCATION `silve-gcp-bucket`').select("url").collect()[0][0]
gold_path = spark.sql('DESCRIBE EXTERNAL LOCATION `gold-gcp-bucket`').select("url").collect()[0][0]
```

```python
def create_Bronze_Schema(environment, path):
    print(f'Using {environment}_Catalog ')
    spark.sql(f'USE CATALOG {environment}_catalog')
    print(f'Creating Bronze Schema in {environment}_Catalog')
    spark.sql(f'CREATE SCHEMA IF NOT EXISTS bronze MANAGED LOCATION "{path}/bronze"')
```
